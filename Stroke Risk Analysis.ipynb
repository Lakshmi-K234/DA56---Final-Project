### Introduction to Stroke Risk Analysis
A **stroke** is a medical condition that occurs when the brain does not receive enough blood flow due to a blockage or bleeding.This sudden interruption stops the supply of oxygen,causing brain cells to get damaged within minutes,leading to lasting brain damage,disability,or death.

The purpose of this analysis is to understand key factors related to stroke and present insights through clear and meaningful data visualization.This project uses Python libraries such as **Pandas**,**Numpy**,**Plotly**,**Matplotlib**,and **Seaborn** to analyze a stroke dataset.

```python
#Create box plot
sns.boxplot(
    x=df["Heart Disease"].map({0:"No",1:"Yes"}),
    y=df["Age"],
    hue=df["Heart Disease"].map({0:"No",1:"Yes"}),
    palette={"No":"#9DB8A0","Yes":"#5A2A4A"},legend=False
)
#Set the title of the plot and axis labels
plt.title("Age Distribution by Heart Disease Status",fontsize=16,pad=20)
plt.xlabel("Heart Disease",fontsize=13,labelpad=15)
plt.ylabel("Age",fontsize=13,labelpad=15)
#Display the plot
plt.show()
```

## Section VII: Socio-Demographic Factors Associated with Stroke Risk
This section examines variations in stroke rates across work types, residential settings, and ever married status. The visualizations reveal demographic differences in stroke risk and show how gender interacts with social factors, indicating that socio-demographic characteristics contribute to observed stroke patterns.

### (i) Stroke Risk Variation by Work Type
• ***Visualization Type:-*** Bar Chart(Matplotlib)  
• ***Purpose:-*** To compare stroke rates across different employment categories and identify work types associated with higher stroke risk.  
• ***Insight:-*** Self-employed individuals show the highest stroke rate,followed by private and government employees,while individuals who never worked exhibit a negligible stroke rate.

```python
#Calculate stroke rate(%) for each work type
stroke_rate_work=(df.groupby("Work Type",observed=True)["Stroke"].mean().mul(100).round(2).sort_values().reset_index())
#Set the figure size
plt.figure(figsize=(8,5))
#Create bar chart
bars=plt.bar(stroke_rate_work["Work Type"],stroke_rate_work["Stroke"],color=["#C94A1A","#E3A14A","#D9822B"])
#Add percentage labels on top of each bar
for bar in bars:
    height=bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2,height,f"{height:.2f}%",ha="center",va="bottom",fontsize=9)
#Set the chart title and axis labels
plt.title("Stroke Rate by Work Type",fontsize=16,pad=20)
plt.xlabel("Work Type",fontsize=13,labelpad=15)
plt.ylabel("Stroke Rate(%)",fontsize=13,labelpad=15)
#Rotate X-axis labels
plt.xticks(rotation=0)
#Add horizontal grid lines
plt.grid(axis="y",linestyle="--",alpha=0.5)
#Display the plot
plt.show()
```

### (ii) Gender-Based Stroke Rate Comparison Across Residence Types
• ***Visualization Type:-*** Grouped Bar Chart(Plotly)  
• ***Purpose:-*** To compare stroke rates between males and females across urban and rural residence types.  
• ***Insight:-*** Stroke rates are consistently higher among males than females in both urban and rural areas,with urban area males showing the highest stroke rate.This suggests gender-based risk differences persist regardless of residence type.

```python
#Group the data by Residence Type and Gender and calculate stroke rate(%)
res_gender=(df.groupby(["Residence Type","Gender"],observed=True)["Stroke"].mean().mul(100).round(2).reset_index())
#Create grouped bar chart
fig=px.bar(res_gender,x="Residence Type",y="Stroke",color="Gender",color_discrete_map={"Female":"#2A9D8F","Male":"#457B9D"},barmode="group",text="Stroke",title="Stroke Rate(%) by Residence Type and Gender",labels={"Stroke":"Stroke Rate(%)"})
#Add percentage labels above bars and format text
fig.update_traces(texttemplate="%{y:.2f}%",textposition="outside")
#Extracts the maximum stroke rate to scale the Y-axis properly
max_val=res_gender["Stroke"].max()
#Center the chart title
fig.update_layout(title=dict(x=0.5,font=dict(size=20),pad=dict(b=20)))
#Set the X-axis label,fontsize and spacing from the axis
fig.update_xaxes(title_text="Residence Type",title_font=dict(size=16),title_standoff=20)
#Adjusts Y-axis range with padding and sets the Y-axis label
fig.update_yaxes(range=[0,max_val + 3],title_text="Stroke Rate(%)",title_font=dict(size=16),title_standoff=20)
#Display the bar chart
fig.show()
```

### (iii) Comparison of Stroke Rates by Ever-Married Status
• ***Visualization Type:-*** Bar Chart(Matplotlib)  
• ***Purpose:-*** To examine the relationship between ever-married status(Yes/No) and stroke occurence by comparing stroke rates across the two groups.  
• ***Insight:-*** Individuals who have ever been married show a higher stroke rate than those who have never been married,indicating a potential association between marital history,and stroke risk,likely influenced by age and related lifestyle factors.

```python
#Calculate stroke rate(%) grouped by Ever Married status
stroke_rate_married=(df.groupby("Ever Married",observed=True)["Stroke"].mean().mul(100).round(2).reset_index())
#Set the figure size
plt.figure(figsize=(6,4))
#Plot bar chart
bars=plt.bar(stroke_rate_married["Ever Married"],stroke_rate_married["Stroke"],color=["#6BAED6","#2171B5"])
#Add values labels above each bar
for bar in bars:
    height=bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2,height,f"{height:.2f}%",ha="center",va="bottom",fontsize=9)
#Add chart title with padding
plt.title("Stroke Rate by Ever Married",fontsize=16,pad=20)
#Label X-axis and Y-axis with adjusting font size and padding
plt.xlabel("Ever Married",fontsize=13,labelpad=15)
plt.ylabel("Stroke Rate(%)",fontsize=13,labelpad=15)
#Set Y-axis limit to add top margin
plt.ylim(0,7.5)
#Add light horizontal gridlines
plt.grid(axis="y",linestyle="--",linewidth=0.6,alpha=0.4)
#Display the plot
plt.show()
```

### (iv) Gender-wise Stroke Rate Across Ever Married Categories
• ***Visualization Type:-*** Heatmap(Seaborn)   
• ***Purpose:-*** To examine how stroke rates vary across gender groups based on ever married(Yes/No).   
• ***Insight:-*** Stroke rates are higher in both genders within one category,with males consistently exhibiting greater stroke prevalence than females,indicating a stronger gender-related riks pattern.

```python
#Calculate stroke rate(%) grouped by Ever Married status
stroke_risk=(df.groupby(["Ever Married","Gender"],observed=True)["Stroke"].mean().mul(100).round(2).unstack(fill_value=0))
#Set the figure size
plt.figure(figsize=(10,6))
#Create heatmap
ax=sns.heatmap(stroke_risk,annot=True,fmt=".1f",cmap="Purples")
#Add value labels above each bar
for t in plt.gca().texts:
    t.set_text(t.get_text() + "%")
#Customize color bar label
cbar=ax.collections[0].colorbar
cbar.set_label("Stroke Rate(%) Range",rotation=90,labelpad=20)    
#Set the chart title
plt.title("Stroke Rate(%) by Gender and Ever Married",fontsize=16,pad=20)
#Label X-axis and Y-axis
plt.xlabel("Gender",fontsize=13,labelpad=15)
plt.ylabel("Ever Married",fontsize=13,labelpad=15)
#Display the plot
plt.show()
```

## Section VIII: Integrated Risk Factors Influencing Stroke
This section analyzes the combined effects of average glucose levels, BMI, cardiovascular risk, and age on stroke occurrence. The visualizations illustrate how stroke rates vary across glucose and BMI categories and how these metabolic factors interact with age groups. The patterns suggest that elevated glucose levels and higher BMI are associated with increased stroke risk, particularly among older age groups and individuals with existing cardiovascular risk.

### (i) Cardiovascular Risk Distribution by BMI and Average Glucose Level
• ***Visualization Type:-*** Scatter Plot(Seaborn)  
• ***Purpose:-*** To examine the relationship between BMI and average glucose level,and to understand how individuals are distributed across different cardiovascular risk levels.     
• ***Insight:-*** Most individuals cluster at BMI 20-40 with glucose levels below 150 mg/dL,indicating lower risk.Higher cardiovascular risk cases appear more frequently at elevated glucose levels(>=180 mg/dL),even across moderate BMI ranges,suggesting glucose level is a stronger risk indicator than BMI alone.

```python
#Set the figure size
plt.figure(figsize=(8,6))
#Create scatterplot
sns.scatterplot(y="BMI (kg/m\u00B2)",x="Avg Glucose Level (mg/dL)",data=df,hue="Cardiovascular Risk",palette={"Low Risk":"#2a9d8f","High Risk":"#e9c46a","Very High Risk":"#9b2226"})
#Add chart title
plt.title("Average Glucose Level vs BMI by Cardiovascular Risk",fontsize=16,pad=20)
#Label the X-axis and Y-axis
plt.xlabel("Average Glucose Level (mg/dL)",fontsize=13,labelpad=15)
plt.ylabel("BMI (kg/m\u00B2)",fontsize=13,labelpad=15)
#Display the scatter plot
plt.show()
```

### (ii) Combined Impact of BMI and Average Glucose Level Categories on Stroke Rate
• ***Visualization Type:-*** Heatmap(Seaborn)  
• ***Purpose:-*** To compare stroke rates(%) across combinations of BMI categories and glucose level categories,highlighting high-risk groups.     
• ***Insight:-*** Stroke rates are highest among overweight and obese individuals with high or prediabetic glucose levels,while underweight and normal glucose groups show minimal stroke occurence.

```python
#Group the data by Glucose Level Category and BMI Category and calculate mean stroke rate(%)
stroke_risk=(df.groupby(["Glucose Level Category","BMI Category"],observed=True)["Stroke"].mean().mul(100).round(2).unstack(fill_value=0))
#Set the figure size
plt.figure(figsize=(10,6))
#Create heatmap
ax=sns.heatmap(stroke_risk,annot=True,fmt=".1f",cmap="viridis")
#Append percentage symbol(%) to all annotated values
for t in plt.gca().texts:
    t.set_text(t.get_text() + "%")
#Customize color bar label
cbar=ax.collections[0].colorbar
cbar.set_label("Stroke Rate(%) Range",rotation=90,labelpad=20)    
#Add the title and axis labels
plt.title("Stroke Rate(%) by Glucose Level Category and BMI Category",fontsize=16,pad=20)
plt.xlabel("BMI Category",fontsize=13,labelpad=15)
plt.ylabel("Glucose Level Category",fontsize=13,labelpad=15)
#Rotate Y-axis labels
plt.yticks(rotation=360)
#Display the heatmap
plt.show()
```

<p style="font-size:16px;text-align:justify">Elevated average glucose levels were strongly associated with stroke occurrence, with diabetic categories consistently exhibiting the highest stroke rates.
